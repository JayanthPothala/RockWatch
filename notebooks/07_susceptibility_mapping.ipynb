# RockWatch - Phase 6: Regional Landslide Susceptibility Mapping
**Project:** Intelligent Landslide Susceptibility Prediction System for Himachal Pradesh  
**Objective:** Generate continuous susceptibility scores and 4-tier hazard classifications across Himachal Pradesh using the champion Gradient Boosting model.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path('.').resolve().parents[0] if Path('.').resolve().name == 'notebooks' else Path('.').resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

preds_df = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'himachal_susceptibility_predictions.csv')
print(f'Total predictions: {len(preds_df):,}')
preds_df.head()
```

## 1. Susceptibility Tier Breakdown

```python
tier_counts = preds_df['susceptibility_class'].value_counts()[['Low', 'Moderate', 'High', 'Very High']]
for tier, count in tier_counts.items():
    print(f'{tier:12s}: {count:5d} ({count/len(preds_df)*100:.1f}%)')
```

## 2. Interactive Susceptibility Map View

```python
tier_colors = {'Low': '#2ca25f', 'Moderate': '#ffbc12', 'High': '#fa8c16', 'Very High': '#d7191c'}

fig, ax = plt.subplots(figsize=(10, 8), dpi=150)
for tier in ['Low', 'Moderate', 'High', 'Very High']:
    sub = preds_df[preds_df['susceptibility_class'] == tier]
    ax.scatter(sub['longitude'], sub['latitude'], c=tier_colors[tier], s=8, alpha=0.6, label=tier)

ax.set_title('Landslide Susceptibility Map of Himachal Pradesh (RockWatch)')
ax.set_xlabel('Longitude (°E)')
ax.set_ylabel('Latitude (°N)')
ax.legend(title='Hazard Tier')
ax.grid(True, linestyle=':', alpha=0.5)
plt.show()
```
